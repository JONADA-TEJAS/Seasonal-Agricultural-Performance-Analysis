## 10. Missing Values Handled

Numerical missing values are replaced using the median because agricultural variables can be skewed and the median is less sensitive to extreme observations.

Categorical missing values are replaced using the mode.

```python
# Identify numerical and categorical columns
numerical_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(exclude=np.number).columns.tolist()

# Fill numerical missing values with median
for column in numerical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].median())

# Fill categorical missing values with mode
for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].mode()[0])

print("Missing values handled successfully.")
print("Remaining missing values:", df.isnull().sum().sum())
```

## 14. Outliers Investigated

Potential outliers are identified using the Interquartile Range (IQR) method.

IQR = Q3 − Q1

Lower Bound = Q1 − 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Outliers are investigated rather than automatically deleted because extreme agricultural observations may represent genuine farms and valid agricultural conditions.

```python
outlier_results = []

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ).sum()

    outlier_results.append({
        "Variable": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_count,
        "Outlier Percentage": round(
            outlier_count / len(df) * 100, 2
        )
    })

outlier_df = pd.DataFrame(outlier_results)

outlier_df.sort_values(
    "Outlier Count",
    ascending=False
)
```

## 26. Student-Designed Analysis 1
### Irrigation Trade-off Analysis

### Analytical Question

Which irrigation method provides the best balance between crop yield, water consumption, and profitability?

A high-yield irrigation method is not automatically the most efficient method if it requires substantially more water.

Therefore, irrigation methods are compared using:
- Average yield
- Average water usage
- Average profit
- Water efficiency

```python
irrigation_analysis = df.groupby("Irrigation_Method").agg(
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Average_Water_Usage=("Water_Used_m3", "mean"),
    Average_Profit=("Profit_INR", "mean"),
    Average_Water_Efficiency=(
        "Water_Efficiency_t_per_1000m3",
        "mean"
    )
).round(3)

irrigation_analysis.sort_values(
    "Average_Water_Efficiency",
    ascending=False
)
```

## 27. Student-Designed Analysis 2
### Crop × Season Profitability Stability

### Analytical Question

Does a crop perform consistently across different seasons?

A crop may have high average profitability overall but perform poorly in a specific season.

Therefore, crop profitability is compared across seasons.

```python
crop_season_profit = pd.pivot_table(
    df,
    values="Profit_INR",
    index="Crop",
    columns="Season",
    aggfunc="mean"
)

crop_season_profit.round(2)
```

```python
plt.figure(figsize=(12, 7))

sns.heatmap(
    crop_season_profit,
    annot=True,
    fmt=".0f",
    cmap="YlGnBu"
)

plt.title("Average Crop Profit by Season")
plt.xlabel("Season")
plt.ylabel("Crop")

plt.tight_layout()
plt.show()
```

## 28. Student-Designed Analysis 3
### State-Level Agricultural Performance

### Analytical Question

Are some states performing better than others in terms of yield, profitability, and resource use?

Geographical differences may be associated with environmental conditions, crop selection, farming practices, and resource availability.

## 29. Student-Designed Analysis 4
### Seasonal Profitability Risk

### Analytical Question

Which season has the highest proportion of farms operating at a loss?

Average profit can hide the distribution of profitable and unprofitable farms.

Therefore, profitability risk is analyzed using the percentage of farms with negative profit.

```python
seasonal_profit_risk = (
    df.groupby("Season")["Profit_INR"]
    .agg(
        Average_Profit="mean",
        Median_Profit="median",
        Loss_Making_Farms=lambda x: (x < 0).sum(),
        Total_Farms="count"
    )
)

seasonal_profit_risk["Loss_Percentage"] = (
    seasonal_profit_risk["Loss_Making_Farms"] /
    seasonal_profit_risk["Total_Farms"] * 100
)

seasonal_profit_risk.round(2)
```

## 30. Student-Designed Analysis 5
### Input Intensity vs Agricultural Performance

### Analytical Question

Do higher fertilizer and pesticide usage levels correspond to higher yield?

This analysis examines whether increasing agricultural input intensity is associated with improved productivity.

```python
input_analysis = df.groupby(
    pd.qcut(
        df["Fertilizer_kg_ha"],
        q=4,
        duplicates="drop"
    )
).agg(
    Average_Fertilizer=("Fertilizer_kg_ha", "mean"),
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Average_Profit=("Profit_INR", "mean"),
    Average_Pesticide=("Pesticide_Litre_ha", "mean")
).round(2)

input_analysis
```

## 31. Economic Efficiency Analysis

Profit margin is calculated to understand how efficiently agricultural revenue is converted into profit.

### Formula

Profit Margin (%) = (Profit / Revenue) × 100

```python
df["Profit_Margin_Percent"] = (
    df["Profit_INR"] /
    df["Revenue_INR"].replace(0, np.nan)
) * 100

df["Profit_Margin_Percent"] = (
    df["Profit_Margin_Percent"]
    .replace([np.inf, -np.inf], np.nan)
)

season_profit_margin = (
    df.groupby("Season")["Profit_Margin_Percent"]
    .mean()
    .round(2)
)

season_profit_margin
```
